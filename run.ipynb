```python
import os
import json

def read_json(json_path):
    """读取 JSON 文件并返回字典"""
    try:
        with open(json_path, 'r', encoding='utf-8') as f:
            return json.load(f)
    except FileNotFoundError:
        print(f"Error: JSON file not found at {json_path}")
        return {}
    except json.JSONDecodeError:
        print(f"Error: Invalid JSON format in {json_path}")
        return {}

def replace_chinese(file_path, translation_dict):
    """将文件中的中文替换为英文"""
    try:
        with open(file_path, 'r', encoding='utf-8') as f:
            content = f.read()
        for zh, en in translation_dict.items():
            content = content.replace(zh, en)
        with open(file_path, 'w', encoding='utf-8') as f:
            f.write(content)
        print(f"Processed: {file_path}")
    except Exception as e:
        print(f"Error processing {file_path}: {str(e)}")

# 读取翻译字典
translation_dict = read_json("web/src/i18n/en.json")

# 检查是否成功加载翻译字典
if not translation_dict:
    print("No translations loaded. Exiting.")
    exit(1)

# 遍历项目文件夹
project_folder = 'task'  # 替换为实际路径
for root, _, files in os.walk(project_folder):
    for file in files:
        # 转换为小写以匹配文件扩展名（不区分大小写）
        if file.lower().endswith(('.js', '.html', '.css', '.go')):
            file_path = os.path.join(root, file)
            replace_chinese(file_path, translation_dict)
```
